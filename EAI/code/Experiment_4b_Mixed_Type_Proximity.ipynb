```python
data = [
    ["A", "Red",   1, "Low",    25],
    ["B", "Blue",  0, "Medium", 30],
    ["C", "Red",   1, "High",   45],
    ["D", "Green", 0, "Low",    35],
    ["E", "Blue",  1, "Medium", 50]
]

ordinal_order = ["Low", "Medium", "High"]

minimum_age = data[0][4]
maximum_age = data[0][4]

for row in data:
    if row[4] < minimum_age:
        minimum_age = row[4]
    if row[4] > maximum_age:
        maximum_age = row[4]

print("Sample Dataset")
print("Object  Colour  Symptom  Level   Age")
for row in data:
    print(row)
```

```python
def calculate_dissimilarity(object1, object2):
    total_difference = 0
    attribute_count = 0

    if object1[1] == object2[1]:
        difference = 0
    else:
        difference = 1

    total_difference = total_difference + difference
    attribute_count = attribute_count + 1

    if not (object1[2] == 0 and object2[2] == 0):
        if object1[2] == object2[2]:
            difference = 0
        else:
            difference = 1

        total_difference = total_difference + difference
        attribute_count = attribute_count + 1

    rank1 = ordinal_order.index(object1[3])
    rank2 = ordinal_order.index(object2[3])

    difference = abs(rank1 - rank2) / (len(ordinal_order) - 1)
    total_difference = total_difference + difference
    attribute_count = attribute_count + 1

    difference = abs(object1[4] - object2[4]) / (maximum_age - minimum_age)
    total_difference = total_difference + difference
    attribute_count = attribute_count + 1

    return total_difference / attribute_count
```

```python
matrix = []

for i in range(len(data)):
    row = []

    for j in range(len(data)):
        if i == j:
            distance = 0
        else:
            distance = calculate_dissimilarity(data[i], data[j])

        row.append(round(distance, 3))

    matrix.append(row)

print("Dissimilarity Matrix")
print("     ", end="")

for object_data in data:
    print("%8s" % object_data[0], end="")

print()

for i in range(len(matrix)):
    print("%5s" % data[i][0], end="")

    for j in range(len(matrix[i])):
        print("%8.3f" % matrix[i][j], end="")

    print()
```
