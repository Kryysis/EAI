```python
import random

random.seed(10)

population = list(range(1, 91))

strata = []

for member in population:
    if member <= 30:
        strata.append("A")
    elif member <= 60:
        strata.append("B")
    else:
        strata.append("C")

print("Population size:", len(population))
```

```python
def random_sampling(population, sample_size):
    sample = random.sample(population, sample_size)
    return sample


random_sample = random_sampling(population, 15)

print("Random Sample:", random_sample)
print("Sample Size:", len(random_sample))
```

```python
def cluster_sampling(population, cluster_size, number_of_clusters):
    clusters = []

    for start in range(0, len(population), cluster_size):
        cluster = population[start:start + cluster_size]
        clusters.append(cluster)

    selected_clusters = random.sample(clusters, number_of_clusters)

    sample = []
    for cluster in selected_clusters:
        for member in cluster:
            sample.append(member)

    return sample


cluster_sample = cluster_sampling(population, 10, 2)

print("Cluster Sample:", cluster_sample)
print("Sample Size:", len(cluster_sample))
```

```python
def stratified_sampling(population, strata, sample_size):
    sample = []
    groups = []

    for group in strata:
        if group not in groups:
            groups.append(group)

    sample_per_group = sample_size // len(groups)

    for group in groups:
        group_members = []

        for i in range(len(population)):
            if strata[i] == group:
                group_members.append(population[i])

        selected_members = random.sample(group_members, sample_per_group)

        for member in selected_members:
            sample.append(member)

    return sample


stratified_sample = stratified_sampling(population, strata, 15)

print("Stratified Sample:", stratified_sample)
print("Sample Size:", len(stratified_sample))
```

```python
random.seed(20)

number_population = []

for i in range(100):
    number = random.randint(10, 100)
    number_population.append(number)

k = 10
sample_size = 20

print("Population:", number_population)
print("Population Size:", len(number_population))
print("Number of samples (k):", k)
print("Size of each sample:", sample_size)
```

```python
def calculate_mean(data):
    return sum(data) / len(data)


def calculate_median(data):
    sorted_data = sorted(data)
    middle = len(sorted_data) // 2

    if len(sorted_data) % 2 == 0:
        median = (sorted_data[middle - 1] + sorted_data[middle]) / 2
    else:
        median = sorted_data[middle]

    return median


def calculate_mode(data):
    frequency = {}

    for value in data:
        if value in frequency:
            frequency[value] = frequency[value] + 1
        else:
            frequency[value] = 1

    highest_frequency = max(frequency.values())

    for value in frequency:
        if frequency[value] == highest_frequency:
            return value


def calculate_range(data):
    return max(data) - min(data)


def calculate_variance(data):
    mean = calculate_mean(data)
    total = 0

    for value in data:
        total = total + (value - mean) ** 2

    return total / len(data)


def calculate_standard_deviation(data):
    variance = calculate_variance(data)
    return variance ** 0.5


def calculate_measures(data):
    measures = [
        calculate_mean(data),
        calculate_median(data),
        calculate_mode(data),
        calculate_range(data),
        calculate_variance(data),
        calculate_standard_deviation(data)
    ]
    return measures


measure_names = ["Mean", "Median", "Mode", "Range", "Variance", "Standard Deviation"]

population_measures = calculate_measures(number_population)

print("Measures of the full population:")
for i in range(len(measure_names)):
    print(measure_names[i], ":", round(population_measures[i], 2))
```

```python
best_sample = []
best_score = None

for sample_number in range(k):
    sample = random.sample(number_population, sample_size)

    sample_measures = calculate_measures(sample)

    total_error = 0

    for i in range(len(population_measures)):
        difference = abs(sample_measures[i] - population_measures[i])
        relative_error = difference / abs(population_measures[i])
        total_error = total_error + relative_error

    score = total_error / len(population_measures)

    print("Sample", sample_number + 1, "Error Score:", round(score, 4))

    if best_score is None or score < best_score:
        best_score = score
        best_sample = sample

print("\nBest Sample:", best_sample)
print("Best Sample Size:", len(best_sample))
print("Best Sample Error Score:", round(best_score, 4))

print("\nMeasures of the best sample:")
best_measures = calculate_measures(best_sample)

for i in range(len(measure_names)):
    print(measure_names[i], ":", round(best_measures[i], 2))
```
