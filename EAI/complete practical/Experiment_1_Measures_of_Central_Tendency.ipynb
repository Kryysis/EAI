# Experiment 1: Measures of Central Tendency

**Aim:** Calculate Mean, Median and Mode for discrete and continuous random samples using NumPy and Pandas. Introduce outliers and observe their effects. Use a sample dataset containing 50–100 instances.

**Note:** This notebook uses 60 observations.

## 1. Import libraries

```python
import numpy as np
import pandas as pd
```

## 2. Create a sample dataset (60 instances)

- **Discrete data:** Random integers from 1 to 10.
- **Continuous data:** Random decimal values from a normal distribution.

Continuous values are rounded to one decimal place so repeated values can be used to find the mode.

```python
np.random.seed(42)

n = 60

discrete = np.random.randint(1, 11, n)
continuous = np.round(np.random.normal(50, 10, n), 1)

data = pd.DataFrame({
    "Discrete": discrete,
    "Continuous": continuous
})

print(data.head())
print("Dataset shape:", data.shape)
```

## Question 1: Calculate measures of central tendency

### A. Using 1D NumPy arrays

A 1D array contains the observations of one variable.

```python
discrete_1d = data["Discrete"].to_numpy()
continuous_1d = data["Continuous"].to_numpy()

print("DISCRETE DATA")
print("Mean:", np.mean(discrete_1d))
print("Median:", np.median(discrete_1d))
print("Mode:", pd.Series(discrete_1d).mode().tolist())

print("\nCONTINUOUS DATA")
print("Mean:", np.mean(continuous_1d))
print("Median:", np.median(continuous_1d))
print("Mode:", pd.Series(continuous_1d).mode().tolist())
```

### B. Using a 2D NumPy array

A 2D array contains both columns. `axis=0` calculates the result separately for each column.

```python
data_2d = data.to_numpy()

print("Mean:", np.mean(data_2d, axis=0))
print("Median:", np.median(data_2d, axis=0))

print("Mode:")
print(data.mode())
```

### C. Using Pandas

A DataFrame is a 2D table. Pandas can calculate these measures directly.

```python
print("Mean:")
print(data.mean())

print("\nMedian:")
print(data.median())

print("\nMode:")
print(data.mode())
```

## Question 2: Introduce outliers and observe their effect

An outlier is an unusually high or low value compared with the other observations. Here, we replace two values with very large values.

```python
data_with_outliers = data.copy()

data_with_outliers.loc[0, "Discrete"] = 100
data_with_outliers.loc[1, "Continuous"] = 200

print(data_with_outliers.head())
```

### Compare the measures before and after adding outliers

```python
print("BEFORE OUTLIERS")
print("Mean:")
print(data.mean())
print("Median:")
print(data.median())
print("Mode:")
print(data.mode())

print("\nAFTER OUTLIERS")
print("Mean:")
print(data_with_outliers.mean())
print("Median:")
print(data_with_outliers.median())
print("Mode:")
print(data_with_outliers.mode())
```

## Question 3: Sample dataset

The dataset created above contains **60 instances** and two variables: one discrete and one continuous. The measures of central tendency were calculated on this dataset, and outliers were added to a copy of it.

## Conclusion

Mean, Median and Mode were calculated for discrete and continuous samples using NumPy and Pandas. The calculations were performed on 1D arrays and a 2D dataset. After adding outliers, the mean increased noticeably, while the median changed only slightly. The mode may or may not change; in this sample, one of the tied discrete modes disappeared.
