# Experiment 6: K-Means Clustering

## Aim
To implement K-Means clustering, a centroid-based clustering technique, using Python.

## Theory
K-Means divides data into **K clusters**. Each cluster is represented by its centroid (the average of the points in that cluster).

### Steps
1. Choose the number of clusters, K.
2. Select initial centroids.
3. Calculate the distance of each point from every centroid.
4. Assign each point to its nearest centroid.
5. Calculate new centroids by taking the average of the points in each cluster.
6. Repeat the assignment and centroid calculation until the centroids stop changing.

In this example, we use two-dimensional data and K = 2.

## Step 1: Create the dataset and choose initial centroids

We use a small dataset containing two groups of points. The first and sixth points are selected as the initial centroids.

```python
import math

data = [
    [1, 2],
    [2, 1],
    [2, 3],
    [3, 2],
    [3, 3],
    [8, 8],
    [9, 8],
    [8, 9],
    [9, 10],
    [10, 9]
]

k = 2

centroids = [data[0], data[5]]

print("Dataset:")
print(data)
print("Initial centroids:", centroids)
```

## Step 2: Implement the K-Means algorithm

The Euclidean distance is used to find the nearest centroid. Points are then assigned to clusters, and new centroids are calculated. The process repeats until the centroids stop changing.

```python
def euclidean_distance(point1, point2):
    distance = math.sqrt(
        (point1[0] - point2[0]) ** 2 +
        (point1[1] - point2[1]) ** 2
    )
    return distance


iteration = 1

while True:
    clusters = []
    for i in range(k):
        clusters.append([])

    for point in data:
        nearest_centroid = 0
        shortest_distance = euclidean_distance(point, centroids[0])

        for i in range(1, k):
            distance = euclidean_distance(point, centroids[i])

            if distance < shortest_distance:
                shortest_distance = distance
                nearest_centroid = i

        clusters[nearest_centroid].append(point)

    new_centroids = []

    for cluster in clusters:
        x_total = 0
        y_total = 0

        for point in cluster:
            x_total = x_total + point[0]
            y_total = y_total + point[1]

        new_x = x_total / len(cluster)
        new_y = y_total / len(cluster)

        new_centroids.append([new_x, new_y])

    if new_centroids == centroids:
        break

    centroids = new_centroids
    iteration = iteration + 1
```

## Step 3: Display the final clusters and centroids

```python
print("Number of iterations:", iteration)

for i in range(k):
    print("\nCluster", i + 1, ":", clusters[i])
    print("Centroid", i + 1, ":", centroids[i])
```

## Conclusion

In this experiment, K-Means clustering was implemented using Python. Each data point was assigned to the cluster with the nearest centroid. The centroids were recalculated repeatedly until they stopped changing, resulting in two final clusters.
