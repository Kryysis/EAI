# Experiment 4(b): Proximity Measure for Mixed-Type Attributes

## Aim
Write a Python script to calculate the dissimilarity between objects in a dataset containing **nominal, asymmetric binary, ordinal, and numerical** attributes. Display the final dissimilarity matrix.

## Formula
The overall dissimilarity is:

$$d(i,j)=\frac{\sum_{f=1}^{p}\delta_{ij}^{(f)}d_{ij}^{(f)}}{\sum_{f=1}^{p}\delta_{ij}^{(f)}}$$

Here, `d` is the dissimilarity for an attribute. `δ` is `1` when the attribute is included and `0` when it is ignored. For an asymmetric binary attribute, a pair where both values are `0` is ignored.

For this practical, we will use a small sample dataset and calculate the dissimilarity using our own Python code.

## 1. Create a sample mixed-type dataset

Each row represents one object:

- **Colour:** Nominal
- **Symptom:** Asymmetric binary (`1` = present, `0` = absent)
- **Level:** Ordinal (`Low < Medium < High`)
- **Age:** Numerical

```python
data = [
    ["A", "Red",   1, "Low",    25],
    ["B", "Blue",  0, "Medium", 30],
    ["C", "Red",   1, "High",   45],
    ["D", "Green", 0, "Low",    35],
    ["E", "Blue",  1, "Medium", 50]
]

ordinal_order = ["Low", "Medium", "High"]

minimum_age = data[0][4]
maximum_age = data[0][4]

for row in data:
    if row[4] < minimum_age:
        minimum_age = row[4]
    if row[4] > maximum_age:
        maximum_age = row[4]

print("Sample Dataset")
print("Object  Colour  Symptom  Level   Age")
for row in data:
    print(row)
```

## 2. Write a function to calculate dissimilarity

We calculate each attribute separately:

1. **Nominal:** `0` if values match, otherwise `1`.
2. **Asymmetric binary:** Ignore the pair if both values are `0`; otherwise `0` if they match and `1` if they differ.
3. **Ordinal:** Convert values to ranks and divide the rank difference by the maximum possible rank difference.
4. **Numerical:** Divide the absolute difference by the range of the numerical attribute.

```python
def calculate_dissimilarity(object1, object2):
    total_difference = 0
    attribute_count = 0

    if object1[1] == object2[1]:
        difference = 0
    else:
        difference = 1

    total_difference = total_difference + difference
    attribute_count = attribute_count + 1

    if not (object1[2] == 0 and object2[2] == 0):
        if object1[2] == object2[2]:
            difference = 0
        else:
            difference = 1

        total_difference = total_difference + difference
        attribute_count = attribute_count + 1

    rank1 = ordinal_order.index(object1[3])
    rank2 = ordinal_order.index(object2[3])

    difference = abs(rank1 - rank2) / (len(ordinal_order) - 1)
    total_difference = total_difference + difference
    attribute_count = attribute_count + 1

    difference = abs(object1[4] - object2[4]) / (maximum_age - minimum_age)
    total_difference = total_difference + difference
    attribute_count = attribute_count + 1

    return total_difference / attribute_count
```

## 3. Calculate and display the dissimilarity matrix

Compare every object with every other object. The diagonal values are `0` because an object has no difference from itself.

```python
matrix = []

for i in range(len(data)):
    row = []

    for j in range(len(data)):
        if i == j:
            distance = 0
        else:
            distance = calculate_dissimilarity(data[i], data[j])

        row.append(round(distance, 3))

    matrix.append(row)

print("Dissimilarity Matrix")
print("     ", end="")

for object_data in data:
    print("%8s" % object_data[0], end="")

print()

for i in range(len(matrix)):
    print("%5s" % data[i][0], end="")

    for j in range(len(matrix[i])):
        print("%8.3f" % matrix[i][j], end="")

    print()
```

## Conclusion

The dissimilarity between objects was calculated using nominal, asymmetric binary, ordinal, and numerical attributes. The final dissimilarity matrix shows the overall difference between every pair of objects. A value closer to **0** indicates that two objects are more similar, while a larger value indicates that they are more different.
