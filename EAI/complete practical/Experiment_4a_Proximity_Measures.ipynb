# Experiment 4(a): Proximity Measures

## Aim
To calculate proximity (similarity or distance) between objects having nominal, binary, numerical and ordinal attributes, and to calculate cosine similarity.

**Note:** Every measure is calculated using our own Python function. No prebuilt similarity or distance functions are used.


## Question (a): Nominal Attributes

For nominal data, compare the attribute values one by one.

- Similarity = Number of matching values / Total number of attributes
- Dissimilarity = 1 - Similarity


```python
object1 = ["Red", "Circle", "Small", "Smooth"]
object2 = ["Red", "Square", "Small", "Rough"]

def nominal_similarity(a, b):
    matches = 0

    for i in range(len(a)):
        if a[i] == b[i]:
            matches = matches + 1

    similarity = matches / len(a)
    return similarity

similarity = nominal_similarity(object1, object2)
dissimilarity = 1 - similarity

print("Nominal Similarity:", similarity)
print("Nominal Dissimilarity:", dissimilarity)
```

## Question (b): Binary Attributes

### i) Symmetric Binary

For symmetric binary attributes, both 0 and 1 are equally important. Count both matching 0s and matching 1s.

### ii) Asymmetric Binary

For asymmetric binary attributes, 1 is more important than 0. Matching 0s are ignored.


```python
A = [1, 0, 1, 0, 1, 0]
B = [1, 0, 0, 0, 1, 1]

def symmetric_binary_similarity(a, b):
    matches = 0

    for i in range(len(a)):
        if a[i] == b[i]:
            matches = matches + 1

    return matches / len(a)

def asymmetric_binary_similarity(a, b):
    both_one = 0
    different = 0

    for i in range(len(a)):
        if a[i] == 1 and b[i] == 1:
            both_one = both_one + 1
        elif a[i] != b[i]:
            different = different + 1

    return both_one / (both_one + different)

print("Symmetric Binary Similarity:", symmetric_binary_similarity(A, B))
print("Asymmetric Binary Similarity:", asymmetric_binary_similarity(A, B))
```

## Question (c): Numerical Attributes

Numerical attributes can be compared using:

- **Manhattan distance:** Add the absolute differences.
- **Euclidean distance:** Take the square root of the sum of squared differences.


```python
x = [1, 2, 3]
y = [2, 4, 4]

def manhattan_distance(a, b):
    distance = 0

    for i in range(len(a)):
        distance = distance + abs(a[i] - b[i])

    return distance

def euclidean_distance(a, b):
    total = 0

    for i in range(len(a)):
        difference = a[i] - b[i]
        total = total + difference * difference

    return total ** 0.5

print("Manhattan Distance:", manhattan_distance(x, y))
print("Euclidean Distance:", euclidean_distance(x, y))
```

## Question (d): Ordinal Attributes

Ordinal values have a meaningful order, such as Low, Medium, High and Very High.

First, convert each category into a rank. Then normalize the rank between 0 and 1:

\(z = (rank - 1) / (number\ of\ categories - 1)\)

Calculate the distance using the normalized ranks.


```python
order = ["Low", "Medium", "High", "Very High"]

object1 = ["Low", "High", "Medium"]
object2 = ["Medium", "Very High", "Medium"]

def ordinal_distance(a, b, order):
    distance = 0

    for i in range(len(a)):
        rank_a = order.index(a[i]) + 1
        rank_b = order.index(b[i]) + 1

        normalized_a = (rank_a - 1) / (len(order) - 1)
        normalized_b = (rank_b - 1) / (len(order) - 1)

        distance = distance + abs(normalized_a - normalized_b)

    return distance

result = ordinal_distance(object1, object2, order)

print("Ordinal Distance:", result)
```

## Question (e): Cosine Similarity

Cosine similarity measures the angle between two numerical vectors. A value closer to 1 means the vectors point in a more similar direction.



```python
vector1 = [1, 2, 3]
vector2 = [2, 1, 3]

def cosine_similarity(a, b):
    dot_product = 0
    sum_a = 0
    sum_b = 0

    for i in range(len(a)):
        dot_product = dot_product + a[i] * b[i]
        sum_a = sum_a + a[i] * a[i]
        sum_b = sum_b + b[i] * b[i]

    length_a = sum_a ** 0.5
    length_b = sum_b ** 0.5

    if length_a == 0 or length_b == 0:
        return 0

    similarity = dot_product / (length_a * length_b)
    return similarity

print("Cosine Similarity:", cosine_similarity(vector1, vector2))
```
